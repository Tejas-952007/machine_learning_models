```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/auto-mpg/auto-mpg.data"

columns = [ "mpg","cylinders","displacement","horsepower","weight","acceleration","model_year","origin","car_name"]

df = pd.read_csv(url,sep=r"\s+",names=columns,na_values="?")
```

```python
print("First 5 rows of dataset:")
print(df.head())

print("\nDataset Information:")
print(df.info())

print("\nStatistical Description:")
print(df.describe())
```

```python
print("\nMissing Values:")
print(df.isnull().sum())

df = df.dropna()

print("\nDataset after removing missing values:")
print(df.shape)
```

```python
X = df[["horsepower"]]
y = df["mpg"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.20,random_state=42)
print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
degrees = [1, 2, 3, 5, 10, 15]
# Lists to store errors
train_mse = []
test_mse = []
train_rmse = []
test_rmse = []
# Dictionary to store trained models
models = {}
```

```python
for degree in degrees:
    print("\n--------------------------------------")
    print("Polynomial Degree:", degree)
    print("--------------------------------------")
    # Create polynomial features
    poly = PolynomialFeatures(degree=degree)
    X_train_poly = poly.fit_transform(X_train)
    X_test_poly = poly.transform(X_test)
    # Create Linear Regression model
    model = LinearRegression()
    # Train model
    model.fit(X_train_poly, y_train)
    # Predictions
    y_train_pred = model.predict(X_train_poly)
    y_test_pred = model.predict(X_test_poly)
    # Calculate MSE
    mse_train = mean_squared_error(y_train, y_train_pred)
    mse_test = mean_squared_error(y_test, y_test_pred)
    # Calculate RMSE
    rmse_train = np.sqrt(mse_train)
    rmse_test = np.sqrt(mse_test)
    # Store values
    train_mse.append(mse_train)
    test_mse.append(mse_test)
    train_rmse.append(rmse_train)
    test_rmse.append(rmse_test)
    models[degree] = (poly, model)
    # Print results
    print("Training MSE :", mse_train)
    print("Testing MSE  :", mse_test)
    print("Training RMSE:", rmse_train)
    print("Testing RMSE :", rmse_test)
```

```python
results = pd.DataFrame({
    "Polynomial Degree": degrees,
    "Training MSE": train_mse,
    "Testing MSE": test_mse,
    "Training RMSE": train_rmse,
    "Testing RMSE": test_rmse
})

print("\n\n==============================================")
print("POLYNOMIAL REGRESSION COMPARISON")
print("==============================================")

print(results.to_string(index=False))
```

```python
best_index = np.argmin(test_mse)

best_degree = degrees[best_index]
best_test_mse = test_mse[best_index]

print("\n==============================================")
print("BEST MODEL")
print("==============================================")

print("Best Polynomial Degree:", best_degree)
print("Minimum Testing MSE:", best_test_mse)
print("Testing RMSE:", test_rmse[best_index])
```

```python
plt.figure(figsize=(8, 5))
plt.scatter( X,y,alpha=0.6)
plt.xlabel("Horsepower")
plt.ylabel("MPG")
plt.title("Horsepower vs MPG")
plt.grid(True)
plt.show()
```

```python
X_sorted = np.sort(X.values, axis=0)
plt.figure(figsize=(12, 8))
plt.scatter(X,y,alpha=0.3,label="Actual Data")
for degree in degrees:
    poly, model = models[degree]
    X_sorted_poly = poly.transform(X_sorted)
    y_sorted_pred = model.predict(X_sorted_poly)
    plt.plot( X_sorted, y_sorted_pred,linewidth=2,label=f"Degree {degree}")
plt.xlabel("Horsepower")
plt.ylabel("MPG")
plt.title("Polynomial Regression for Different Degrees")
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(9, 6))
plt.plot(degrees,train_mse,marker="o",label="Training MSE")
plt.plot(degrees,test_mse,marker="o",label="Testing MSE")
plt.xlabel("Polynomial Degree")
plt.ylabel("Mean Squared Error")
plt.title("Training Error vs Testing Error")
plt.xticks(degrees)
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.plot(degrees,train_rmse,marker="o",label="Training RMSE")
plt.plot(degrees,test_rmse,marker="o",label="Testing RMSE")
plt.xlabel("Polynomial Degree")
plt.ylabel("RMSE")
plt.title("Training RMSE vs Testing RMSE")
plt.xticks(degrees)
plt.legend()
plt.grid(True)
plt.show()
```

```python
print("\n==============================================")
print("MODEL ANALYSIS")
print("==============================================")
for i, degree in enumerate(degrees):
    difference = test_mse[i] - train_mse[i]
    print(f"\nDegree {degree}")
    print("Training MSE:", train_mse[i])
    print("Testing MSE :", test_mse[i])
    if degree == 1:
        print("Observation: Linear model may underfit nonlinear relationship.")
    elif test_mse[i] == min(test_mse):
        print("Observation: Good fit - testing error is minimum.")
    elif difference > test_mse[i] * 0.5:
        print("Observation: Possible overfitting.")
    else:
        print("Observation: Reasonable model complexity.")
```
